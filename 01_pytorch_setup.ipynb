```python

import torch

print("PyTorch version:", torch.__version__)
print("GPU available:", torch.cuda.is_available())

x = torch.tensor([1.0, 2.0, 3.0])
print("My first tensor:", x)
print("Tensor multiplied by 2:", x * 2)


a = torch.tensor(10.0, requires_grad=True)

b = a**2

b.backward()

print("Value of b:", b.item())
print("Gradient of b:", a.grad.item())
```

```python

import torch

# Create a one-dimensional tensor
a = torch.tensor([10, 20, 30])

# Create a two-dimensional tensor
b = torch.tensor([
    [1, 2, 3],
    [4, 5, 6]
])

print("a:", a)
print("Shape of a:", a.shape)
print("Dimensions of a:", a.ndim)

print("b:", b)
print("Shape of b:", b.shape)
print("Dimensions of b:", b.ndim)
```

```python

import torch

a = torch.tensor([10, 20, 30])
b = torch.tensor([1, 2, 3])

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
```
